# Pick DEM bounds interactively

Draw a rectangle on a map instead of typing `DEM_BOUNDS_WGS84` coordinates by hand.
This is Jupyter-only — for a plain `.py` script or the CLI, pass the bounds
as a literal `(min_lon, min_lat, max_lon, max_lat)` tuple.

Requires the `notebook` extra:
```bash
pip install hydroflow[notebook]
```

````python
from hydroflow.utils.notebook_map import pick_bounds_map, get_drawn_bounds
````

Display the map, then use the rectangle tool in the draw toolbar (left side)
to draw a box around your basin of interest.

````python
m = pick_bounds_map(center=(27.7, 85.3), zoom=9)
m
````

After drawing a rectangle above, run this cell to extract the bounds.

````python
bounds = get_drawn_bounds(m)
bounds   # (min_lon, min_lat, max_lon, max_lat)
````

Feed the drawn bounds straight into a `Config` and run the pipeline — no
manual coordinate typing needed. See [Examples](https://pyhydroflow.readthedocs.io/en/latest/examples/)
for what to do next (boundary conditions, elevation-based Manning's n, VSA-OPM runoff, ...).

````python
from hydroflow import Config, run_pipeline, plot_watershed

cfg = Config(
    DEM_BOUNDS_WGS84=bounds,
    DEM_SOURCE="nasadem",
    OUTPUT_DIR="results/",
    OUTPUT_POINT=(27.7, 85.3),   # (lat, lon) — set this to your actual outlet
    TARGET_CRS_EPSG="EPSG:32645",
    GEE_PROJECT="your-gee-project",
)
cfg.update_output_paths()

out = run_pipeline(cfg, stages=("process_dem",))
fig, ax = plot_watershed(out)
````
